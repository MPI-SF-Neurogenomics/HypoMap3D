```python
import warnings
import os 
from pathlib import Path

import zarr
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import to_hex
import matplotlib.colors as mcolors
from matplotlib import gridspec
import seaborn as sns
from dotenv import dotenv_values

from hm3d_utils.plotting import plot_clusters_collapsed
```

```python
# Setting up filepaths from config
CONFIG = dotenv_values("config.env") 
DATADIR = Path(CONFIG['DATADIR'])

OUTDIR = DATADIR / "publication" / "Fig_04"
os.makedirs(OUTDIR, exist_ok=True)

ST_FP = DATADIR / "9_objects" / "spatial" / 'hypomap3d_spatial_voxel_level.zarr'
```

```python
store = zarr.open(ST_FP, mode='r')
```

## ARH cell types

```python
color_mapping_arh = {
    'Gl4-212': "#1f77b4", 
    'Gl4-213': "#ff7f0e",
}
```

```python
# Coronal view - sum over AP range, plot [DV, LM]
plot_clusters_collapsed(store, 
    color_mapping_arh.keys(), 
    slice_range=(85, 135),
    thresholds     = [0.6, 0.8],    
    regions=['HY', 'ARH', 'DMH', 'VMH', 'LHA','TU'], 
    view='coronal',
    alphas         = [0.2, 0.5],
    linestyles = ['dashed', 'solid'],
    linewidths = [0.6,0.6],  
    axis0_range    = (60, 140)  ,
    axis1_range    = (60, 110)  ,
    figsize        = (4,4),
    label_fontsize = 11,
   region_label_fontsize = 11,
   title="Coronal",
   color_mapping = color_mapping_arh,
   save_path = OUTDIR / "arh_coronal.pdf"
)
```

```python
# Coronal view - sum over AP range, plot [DV, LM]
plot_clusters_collapsed(store, 
    color_mapping_arh.keys(), 
    slice_range=(85, 135),
    thresholds     = [0.6, 0.8],
    regions= ['HY', 'ARH', 'DMH', 'VMH','AHN','PH', "PVp", "PMd", "TMd", "SUM"],
    view='sagittal',
    alphas         = [0.2, 0.5],
    linestyles = ['dashed', 'solid'],
    linewidths = [0.6,0.6],  
    axis0_range    = (60, 140)  ,
    axis1_range    = (85, 135)  ,
    figsize        = (4,4),
    label_fontsize = 11  ,
   region_label_fontsize = 11,
   color_mapping = color_mapping_arh,
   title="Sagittal",
   save_path = OUTDIR / "arh_sagittal.pdf"
)
```
