1. Subset AnnData to only non control samples

```python
from pathlib import Path
import os 

import scanpy as sc
import numpy as np
from dotenv import dotenv_values
```

```python
CONFIG = dotenv_values("config.env") 
DATADIR = Path(CONFIG['DATADIR'])
DATASETS_DIR = DATADIR / "4_HY" / 'all_HY' / "datasets"
OUTDIR = DATADIR / "6_NC_projection" 
os.makedirs(OUTDIR, exist_ok=True)

ADATA_FP = DATASETS_DIR / 'HY_raw_embed.h5ad'

OUTFILE = OUTDIR / 'non_control.h5ad'
```

```python
adata = sc.read_h5ad(ADATA_FP, backed='r')
adata = adata[~adata.obs['IsControl']].to_memory()

# Check dtype and convert if it's an integer to avoid rounding errors during normalization
if np.issubdtype(adata.X.dtype, np.integer):
    adata.X = adata.X.astype(np.float32)
```

```python
adata.write_h5ad(OUTFILE)
```
