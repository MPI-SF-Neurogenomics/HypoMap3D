```python
from pathlib import Path

import anndata as ad

from dotenv import dotenv_values
```

```python
# Load config file agnostic to project or notebook work directory
config_path = Path("config.env")
if not config_path.exists():
    config_path = Path("../../config.env")
CONFIG = dotenv_values(config_path)
assert CONFIG != {}, 'Configuration is empty - check the filepath.'

DATADIR = Path(CONFIG['DATADIR'])
```

```python
# Load single-cell control data
fp_sc_control = DATADIR / "9_objects/singlecell/hypomap3d_singlecell_control.h5ad"
scdata = ad.read_h5ad(fp_sc_control)
scdata
```

```python
hierarchy = scdata.obs[
    [
        'L1_ID', 'L2_ID', 'L3_ID', 'L4_ID', 'L5_ID',
        'L1_descr', 'L2_descr', 'L3_descr', 'L4_descr', 'L5_descr',
        'L1_abc_like', 'L2_abc_like', 'L3_abc_like', 'L4_abc_like', 'L5_abc_like',
    ]
].drop_duplicates(ignore_index=True)

hierarchy
```

```python
# Construct a flattened dictionary of all clusters

cluster_taxonomy = {}

# Iterate over each row in the hierarchy and populate the cluster dictionary
for _, row in hierarchy.iterrows():
    for level in range(1, 6):
        cluster_id = row[f'L{level}_ID']
        if cluster_id not in cluster_taxonomy:
            cluster_taxonomy[cluster_id] = {
                'descr': row[f'L{level}_descr'],
                'abc_like': row[f'L{level}_abc_like'],
                'parent': row[f'L{level-1}_ID'] if level > 1 else 'root',
                'path': [row[f'L{i}_ID'] for i in range(1, level+1)],
                'children': [],
                'sisters': []
            }

# Populate children and sisters
for cluster_id, cluster_info in cluster_taxonomy.items():
    parent_id = cluster_info['parent']
    if parent_id in cluster_taxonomy:
        cluster_taxonomy[parent_id]['children'].append(cluster_id)
        cluster_taxonomy[cluster_id]['sisters'] = [
            sib for sib in cluster_taxonomy[parent_id]['children'] if sib != cluster_id
        ]

# Update sister of 'root' parents
for cluster_id, cluster_info in cluster_taxonomy.items():
    if cluster_info['parent'] == 'root':
        cluster_info['sisters'] = [
            sib for sib in cluster_taxonomy if cluster_taxonomy[sib]['parent'] == 'root' and sib != cluster_id
        ]

# Add n_sisters, n_children
for cluster_id, cluster_info in cluster_taxonomy.items():
    cluster_info['n_sisters'] = len(cluster_info['sisters'])
    cluster_info['n_children'] = len(cluster_info['children'])

cluster_taxonomy
```

```python
import json

outfile = DATADIR / "9_objects/taxonomy.json"

with open(outfile, 'w') as f:
    json.dump(cluster_taxonomy, f, indent=2)
```
