Notebook goals:
- Concat a selection of pucks spanning the AP HY range from the [Langlieb et al publication](https://www.nature.com/articles/s41586-023-06818-7) into one anndata object
- Add refined CCF coordinates from our own CCF registration
- Add some ABC metadata to .obs
- Prefilter to HY bounding box to save some disk space and memory

__If you would like to rerun this notebook you would need access to our updated registrations which are currently not publicly available.
Hence, please contact us if that is the case.__

There are some other requirements that currently require manual labor (might be refactored later):
- The raw Langlieb data itself. For this, download the pucks from [here](https://drive.google.com/drive/folders/1ajep-hCLW1BJ-ljS_3ApjgpVV4uELpFg) into your PUCK_DIR (you'll be needing pucks 18-74)
- The ABC structure ontology/tree from [here](https://community.brain-map.org/t/allen-mouse-ccf-accessing-and-using-related-data-and-tools/359) under "how do I download the structure ontology/tree", then get the CSV and put into your DOWNLOAD_DIR
- Some serious MEM as a lot of data has to be loaded and concatenated (might be refactored later)

```python
from pathlib import Path

from dotenv import dotenv_values

from hm3d_utils.spatial import create_full_macosko, BoundingBox3D
```

```python
# Settuping up filepaths from config
CONFIG = dotenv_values("config.env") 
DATADIR = Path(CONFIG['DATADIR'])

OUTDIR = DATADIR / "7_spatial" / "raw"
DOWNLOAD_DIR = OUTDIR / "Langlieb"
DOWNLOAD_DIR.mkdir(exist_ok=True, parents=True) 

CCF_ANNOTATION_FP = DATADIR / "7_spatial" / 'abc' / '25um_annotation_mtx.nrrd'
REGISTRATION_DIR = DOWNLOAD_DIR / "registration"
PUCK_DIR = DOWNLOAD_DIR / "All_Pucks_h5ad"
QUERY_FP = DOWNLOAD_DIR / "query.csv"

BBOX_FP = DATADIR / "7_spatial" / 'bounding_box_ccf_mm.json'

OUTFILE = OUTDIR / "Langlieb_spatial.h5ad"
```

```python
bbox_ccf = BoundingBox3D.load(BBOX_FP) # passing this bbox will filter the data to the HY already
```

```python
#only_load = [52, 54] # testing

only_load = [] # default

mac = create_full_macosko(puck_dir=str(PUCK_DIR), #the function currently only accepts str
                          ccf_annotation_file=str(CCF_ANNOTATION_FP),
                          registration_dir=str(REGISTRATION_DIR),
                          query_file_path = str(QUERY_FP),
                          only_load = only_load,
                          save = False,
                          save_path = str(OUTFILE),
                          bbox_ccf=bbox_ccf)
mac
```
